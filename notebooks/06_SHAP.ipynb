```python
import shap
import numpy as np
import pandas as pd
import joblib
```

```python
pipeline = joblib.load('/Workspace/Users/ajiboyeniola@gmail.com/lead-scoring/models/lead_scoring_model.pkl')
```

```python
# Extract the actual pipeline from the dict
actual_pipeline = pipeline['model']

preprocessor = actual_pipeline.named_steps['preprocessing']
model = actual_pipeline.named_steps['model']
```

```python
# Extract column names from the preprocessor transformers
robust_cols = preprocessor.transformers_[0][2]
oe_cols = preprocessor.transformers_[1][2]
ohe_cols = preprocessor.transformers_[2][2]

robust_feature_names = robust_cols
oe_feature_names = oe_cols
ohe_feature_names = preprocessor.named_transformers_['ohe'].get_feature_names_out(ohe_cols).tolist()
```

```python
all_input_cols = list(robust_cols) + list(oe_cols) + list(ohe_cols)

# Get passthrough columns from the preprocessor's feature names
if hasattr(preprocessor, 'feature_names_in_'):
    all_input_features = list(preprocessor.feature_names_in_)
    passthrough_cols = [col for col in all_input_features if col not in all_input_cols]
else:
    # Fallback: no passthrough columns
    passthrough_cols = []
```

```python
feature_names = robust_feature_names + oe_feature_names + ohe_feature_names + passthrough_cols
```

```python
print(f"Total features after preprocessing: {len(feature_names)}")
print(feature_names)
```
